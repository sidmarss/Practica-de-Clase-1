# Entregable 1. Definición del Problema

### 1. Representación de una configuración de los 8 satélites y su información

Para convertir el problema físico en un modelo computacional tratable, se representa la constelación de los 8 satélites como un vector o cromosoma de coordenadas geográficas sobre la superficie esférica de la Tierra.

* **Estructura del Individuo:** Cada configuración (individuo) es una lista de 8 pares de valores reales que corresponden a la latitud ($\phi$) y longitud ($\lambda$) de cada satélite:

$$\text{Individuo} = [(\phi_1, \lambda_1), (\phi_2, \lambda_2), \dots, (\phi_8, \lambda_8)]$$


* **Dominio espacial:** Las latitudes se restringen al rango $[-80^\circ, 80^\circ]$ y las longitudes a $[-180^\circ, 180^\circ]$ para evitar singularidades en los polos y mantener una distribución orbital realista en el modelo simplificado.


### 2. Significado y determinación de una región cubierta

Una región de interés se considera **cubierta** de manera efectiva si al menos un satélite de la constelación se encuentra dentro de su radio de cobertura operativo.

* **Cálculo matemático:** Para determinar si un satélite en $(\phi_s, \lambda_s)$ cubre una región en $(\phi_r, \lambda_r)$, el sistema calcula la distancia geodésica sobre la superficie terrestre utilizando la **fórmula de Haversine**, la cual asume un radio medio de la Tierra $R_T \approx 6,371\text{ km}$.
* **Umbral de cobertura:** Se establece un radio de observación estándar de **$300\text{ km}$**. Si la distancia geodésica $d \le 300\text{ km}$, la región queda formalmente registrada como cubierta.


### 3. Función de evaluación (Fitness) y consideración de prioridades

La calidad de una configuración se mide mediante una función de aptitud (*fitness*) multiobjetivo integrada que premia la cobertura de las regiones ponderada por su importancia y penaliza el incumplimiento de restricciones.

* **Pesos de Prioridad ($w_j$):** Las 6 regiones se clasifican según su importancia inicial asignándoles un peso numérico:
* **Alta (Prioridad 3):** Península de Yucatán y Amazonía Brasileña ($w = 3$).
* **Media (Prioridad 2):** California, Región del Sahel y Japón ($w = 2$).
* **Normal (Prioridad 1):** Interior de Australia ($w = 1$).


* **Fórmula de Fitness:**

$$\text{Fitness} = \left( \sum_{j=1}^{6} w_j \cdot \mathbb{I}(\text{Región}_j \text{ cubierta}) \times 10 \right) - \text{Penalizaciones}$$



*(Donde $\mathbb{I}$ es una función indicatriz que vale 1 si la región está cubierta y 0 en caso contrario).*


### 4. Condiciones que hacen que una configuración sea inválida

Una solución se considera **inválida** o penalizada cuando vulnera las restricciones físicas del espacio aéreo/orbital impuestas por el escenario:

* **Invasión de zonas de desechos espaciales:** Si la posición de un satélite se encuentra a una distancia angular menor o igual al radio de restricción definido para el Atlántico Norte ($12^\circ$), Pacífico Sur ($10^\circ$) o Región Polar Norte ($8^\circ$).
* **Interferencia con objetos ocupados:** Si un satélite infringe el radio de exclusión de $5^\circ$ correspondiente al **Objeto A** o al **Objeto B**.
* *Tratamiento computacional:* En lugar de descartar drásticamente al individuo, el sistema aplica una **penalización económica severa** en la función de fitness (restando puntos significativos por cada infracción), guiando al algoritmo evolutivo a descartar naturalmente estas zonas en generaciones posteriores.


### 5. Estrategia general del algoritmo evolutivo

Para explorar y optimizar el espacio de soluciones, el sistema implementa un **Algoritmo Evolutivo (AE)** generacional con los siguientes operadores:

1. **Inicialización:** Se genera una población inicial de individuos cuyas posiciones de satélites se validan para evitar zonas de exclusión desde el inicio.
2. **Selección:** Se utiliza el método de **torneo binario**, seleccionando a los mejores individuos compitiendo en parejas para asegurar presión selectiva hacia mejores coberturas.
3. **Cruce (Crossover):** Se aplica un cruce de un punto entre los cromosomas de los padres para combinar distribuciones espaciales prometedoras.
4. **Mutación:** Se introduce una pequeña perturbación gaussiana estocástica en las coordenadas $(\phi, \lambda)$ de los satélites con una probabilidad controlada ($p_m$), permitiendo explorar nuevas ubicaciones geográficas cercanas.
5. **Elitismo:** Se preserva de manera intacta al mejor individuo de cada generación hacia la siguiente, garantizando que la calidad de la solución óptima nunca decaiga.


### 6. Parámetros del sistema y valores iniciales

Los parámetros principales que rigen el comportamiento del algoritmo y su justificación son:

* **Tamaño de población (`pop_size = 80`):** Suficientemente amplio para mantener diversidad genética y explorar múltiples sectores geográficos del globo sin ralentizar excesivamente el cómputo.
* **Número de generaciones (`generations = 120`):** Margen adecuado para permitir la convergencia estable del fitness global.
* **Tasa de mutación (`mutation_rate = 0.25`):** Permite un balance óptimo entre la explotación de configuraciones exitosas y la exploración de nuevas áreas libres de restricciones.
* **Número de satélites:** Fijo en $8$ unidades, conforme a la limitación de recursos de la organización.


### 7. Criterios para considerar una solución suficientemente buena

Una solución obtenida por el sistema se considerará exitosa y óptima para el escenario cuando cumpla simultáneamente con lo siguiente:

1. **Cero violaciones de restricciones:** El conteo de infracciones en zonas de desechos y objetos ocupados debe ser exactamente $0$.
2. **Cobertura máxima ponderada:** Lograr cubrir el 100% de las regiones de alta y media prioridad (Yucatán, Amazonía, California, Sahel y Japón), maximizando opcionalmente el alcance hacia la región de prioridad normal (Australia) según la geometría espacial permitida.
3. **Estabilidad de convergencia:** La gráfica de evolución del fitness muestre una asíntota horizontal plana en las últimas generaciones, evidenciando que el algoritmo ha alcanzado su límite de mejora óptimo.

# **Entregable 2. Implementación funcional**

```python
import random
import math
import numpy as np
import plotly.graph_objects as go

# ==========================================
# 1. DEFINICIÓN DEL ESCENARIO Y CONSTANTES
# ==========================================
EARTH_RADIUS_KM = 6371.0
OBSERVATION_RADIUS_KM = 300.0

# Regiones de interés con sus prioridades (pesos)
REGIONS = [
    {"name": "Península de Yucatán", "lat": 20.5, "lon": -89.0, "priority": 3},
    {"name": "Amazonía Brasileña", "lat": -3.5, "lon": -62.0, "priority": 3},
    {"name": "California, EE.UU.", "lat": 36.8, "lon": -119.4, "priority": 2},
    {"name": "Región del Sahel", "lat": 15.0, "lon": 0.0, "priority": 2},
    {"name": "Japón", "lat": 36.0, "lon": 138.0, "priority": 2},
    {"name": "Interior de Australia", "lat": -25.0, "lon": 133.0, "priority": 1},
]

# Zonas restringidas por desechos espaciales
DEBRIS_ZONES = [
    {"name": "Atlántico Norte", "lat": 45.0, "lon": -35.0, "radius_deg": 12.0},
    {"name": "Pacífico Sur", "lat": -30.0, "lon": -120.0, "radius_deg": 10.0},
    {"name": "Región Polar Norte", "lat": 78.0, "lon": 30.0, "radius_deg": 8.0},
]

# Posiciones ocupadas por objetos existentes
BLOCKED_OBJECTS = [
    {"name": "Objeto A", "lat": 10.0, "lon": 45.0, "radius_deg": 5.0},
    {"name": "Objeto B", "lat": -15.0, "lon": 110.0, "radius_deg": 5.0},
]

# ==========================================
# 2. FUNCIONES GEODÉSICAS Y DE RESTRICCIÓN
# ==========================================
def haversine_distance(lat1, lon1, lat2, lon2):
    dlat = math.radians(lat2 - lat1)
    dlon = math.radians(lon2 - lon1)
    a = (math.sin(dlat / 2) ** 2 +
         math.cos(math.radians(lat1)) * math.cos(math.radians(lat2)) * math.sin(dlon / 2) ** 2)
    c = 2 * math.asin(math.sqrt(a))
    return EARTH_RADIUS_KM * c

def angular_distance(lat1, lon1, lat2, lon2):
    phi1, lambda1 = math.radians(lat1), math.radians(lon1)
    phi2, lambda2 = math.radians(lat2), math.radians(lon2)
    cos_angle = (math.sin(phi1) * math.sin(phi2) +
                 math.cos(phi1) * math.cos(phi2) * math.cos(lambda1 - lambda2))
    cos_angle = max(-1.0, min(1.0, cos_angle))
    return math.degrees(math.acos(cos_angle))

def check_constraints(lat, lon):
    """Verifica si un punto infringe zonas de desechos o objetos ocupados."""
    for debris in DEBRIS_ZONES:
        if angular_distance(lat, lon, debris["lat"], debris["lon"]) <= debris["radius_deg"]:
            return False
    for obj in BLOCKED_OBJECTS:
        if angular_distance(lat, lon, obj["lat"], obj["lon"]) <= obj["radius_deg"]:
            return False
    return True

# ==========================================
# 3. FUNCIÓN DE EVALUACIÓN (FITNESS)
# ==========================================
def evaluate_individual(individual):
    penalty = 0
    valid_satellites = 0

    for sat in individual:
        lat, lon = sat
        if check_constraints(lat, lon):
            valid_satellites += 1
        else:
            penalty += 100  # Fuerte penalización por incumplir restricciones

    covered_score = 0
    region_status = []

    for reg in REGIONS:
        is_covered = False
        for sat in individual:
            if haversine_distance(sat[0], sat[1], reg["lat"], reg["lon"]) <= OBSERVATION_RADIUS_KM:
                is_covered = True
                break
        if is_covered:
            covered_score += reg["priority"] * 10
            region_status.append(True)
        else:
            region_status.append(False)

    fitness = covered_score - penalty + (valid_satellites * 0.5)
    return fitness, region_status, penalty

# ==========================================
# 4. MOTOR DEL ALGORITMO EVOLUTIVO
# ==========================================
class EvolutionaryConstellationOptimizer:
    def __init__(self, pop_size=80, generations=120, mutation_rate=0.25):
        self.pop_size = pop_size
        self.generations = generations
        self.mutation_rate = mutation_rate
        self.num_satellites = 8

    def generate_random_satellite(self):
        while True:
            lat = random.uniform(-80, 80)
            lon = random.uniform(-180, 180)
            if check_constraints(lat, lon):
                return (lat, lon)

    def initialize_population(self):
        return [[self.generate_random_satellite() for _ in range(self.num_satellites)] for _ in range(self.pop_size)]

    def selection(self, population, fitnesses):
        selected = []
        for _ in range(len(population)):
            i, j = random.sample(range(len(population)), 2)
            selected.append(population[i] if fitnesses[i] > fitnesses[j] else population[j])
        return selected

    @staticmethod
    def crossover(p1, p2):
        point = random.randint(1, len(p1) - 1)
        return p1[:point] + p2[point:], p2[:point] + p1[point:]

    def mutate(self, individual):
        mutated = []
        for sat in individual:
            if random.random() < self.mutation_rate:
                lat = max(-80, min(80, sat[0] + random.gauss(0, 5)))
                lon = max(-180, min(180, sat[1] + random.gauss(0, 10)))
                if check_constraints(lat, lon):
                    mutated.append((lat, lon))
                else:
                    mutated.append(sat)
            else:
                mutated.append(sat)
        return mutated

    def optimize(self):
        population = self.initialize_population()
        best_ever = None
        best_fitness = -float('inf')
        history = []

        for gen in range(self.generations):
            evaluated = [evaluate_individual(ind) for ind in population]
            fitnesses = [res[0] for res in evaluated]

            max_fit_idx = np.argmax(fitnesses)
            if fitnesses[max_fit_idx] > best_fitness:
                best_fitness = fitnesses[max_fit_idx]
                best_ever = population[max_fit_idx]

            history.append(best_fitness)

            next_generation = [best_ever]  # Elitismo
            selected = self.selection(population, fitnesses)

            while len(next_generation) < self.pop_size:
                p1, p2 = random.sample(selected, 2)
                c1, c2 = self.crossover(p1, p2)
                next_generation.append(self.mutate(c1))
                if len(next_generation) < self.pop_size:
                    next_generation.append(self.mutate(c2))

            population = next_generation

        return best_ever, best_fitness, history

# Ejecución de la optimización
optimizer = EvolutionaryConstellationOptimizer()
best_constellation, best_fit, fitness_history = optimizer.optimize()
print(f"Optimización completada con éxito. Mejor Aptitud: {best_fit:.2f}")

# ==========================================
# 5. VISUALIZACIÓN 3D INTERACTIVA (PLOTLY)
# ==========================================
def plot_interactive_earth(constellation):
    def latlon_to_cartesian(lat, lon, radius=EARTH_RADIUS_KM):
        rad_lat = np.radians(lat)
        rad_lon = np.radians(lon)
        return radius * np.cos(rad_lat) * np.cos(rad_lon), radius * np.cos(rad_lat) * np.sin(rad_lon), radius * np.sin(rad_lat)

    data = []

    # Esfera de la Tierra
    u, v = np.linspace(0, 2 * np.pi, 50), np.linspace(0, np.pi, 50)
    xe, ye, ze = EARTH_RADIUS_KM * np.outer(np.cos(u), np.sin(v)), EARTH_RADIUS_KM * np.outer(np.sin(u), np.sin(v)), EARTH_RADIUS_KM * np.outer(np.ones(np.size(u)), np.cos(v))
    data.append(go.Surface(x=xe, y=ye, z=ze, colorscale=[[0, 'rgba(0, 119, 190, 0.25)'], [1, 'rgba(0, 119, 190, 0.25)']], showscale=False, name="Tierra"))

    # Regiones
    rx, ry, rz, r_txt = [], [], [], []
    for reg in REGIONS:
        x, y, z = latlon_to_cartesian(reg["lat"], reg["lon"])
        rx.append(x); ry.append(y); rz.append(z)
        r_txt.append(f"<b>{reg['name']}</b><br>Prioridad: {reg['priority']}")
    data.append(go.Scatter3d(x=rx, y=ry, z=rz, mode='markers+text', marker=dict(size=7, color='limegreen'), text=[r['name'] for r in REGIONS], textposition="top center", textfont=dict(color='white', size=10), hovertext=r_txt, hoverinfo="text", name="Regiones"))

    # Desechos
    dx, dy, dz, d_txt = [], [], [], []
    for deb in DEBRIS_ZONES:
        x, y, z = latlon_to_cartesian(deb["lat"], deb["lon"])
        dx.append(x); dy.append(y); dz.append(z)
        d_txt.append(f"<b>Desechos: {deb['name']}</b>")
    data.append(go.Scatter3d(x=dx, y=dy, z=dz, mode='markers+text', marker=dict(size=9, color='red', symbol='x'), text=[d['name'] for d in DEBRIS_ZONES], textposition="top center", textfont=dict(color='white', size=10), hovertext=d_txt, hoverinfo="text", name="Desechos"))

    # Objetos Ocupados
    ox, oy, oz, o_txt = [], [], [], []
    for obj in BLOCKED_OBJECTS:
        x, y, z = latlon_to_cartesian(obj["lat"], obj["lon"])
        ox.append(x); oy.append(y); oz.append(z)
        o_txt.append(f"<b>{obj['name']}</b>")
    data.append(go.Scatter3d(x=ox, y=oy, z=oz, mode='markers+text', marker=dict(size=9, color='orange', symbol='diamond'), text=[o['name'] for o in BLOCKED_OBJECTS], textposition="top center", textfont=dict(color='white', size=10), hovertext=o_txt, hoverinfo="text", name="Objetos Ocupados"))

    # Satélites
    sx, sy, sz, s_txt = [], [], [], []
    for i, sat in enumerate(constellation):
        x, y, z = latlon_to_cartesian(sat[0], sat[1], radius=EARTH_RADIUS_KM + 400)
        sx.append(x); sy.append(y); sz.append(z)
        s_txt.append(f"<b>Satélite {i+1}</b><br>Lat: {sat[0]:.2f}°<br>Lon: {sat[1]:.2f}°")
    data.append(go.Scatter3d(x=sx, y=sy, z=sz, mode='markers+text', marker=dict(size=10, color='cyan', symbol='square'), text=[f"Sat-{i+1}" for i in range(len(constellation))], textposition="top center", textfont=dict(color='cyan', size=11), hovertext=s_txt, hoverinfo="text", name="Satélites"))

    layout = go.Layout(
        title=dict(text="<b>AstroShield — Implementación Funcional 3D</b>", font=dict(color='white', size=16)),
        paper_bgcolor='black', plot_bgcolor='black',
        scene=dict(
            xaxis=dict(title='X (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            yaxis=dict(title='Y (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            zaxis=dict(title='Z (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            aspectmode='cube'
        ),
        legend=dict(x=0, y=0.9, font=dict(color='white'), bgcolor='rgba(0,0,0,0.5)'),
        margin=dict(l=0, r=0, b=0, t=40)
    )

    fig = go.Figure(data=data, layout=layout)
    fig.show()

plot_interactive_earth(best_constellation)
```

Gestión de Restricciones Geométricas: Utiliza la fórmula de Haversine y distancia angular esférica para validar que ningún satélite invada las zonas de desechos espaciales ni los objetos ocupados ($A$ y $B$).Función de Aptitud (Fitness): Calcula el nivel de cobertura ponderada de las 6 regiones objetivo aplicando penalizaciones dinámicas estrictas ante cualquier infracción de restricciones espaciales.Motor Evolutivo Generacional: Implementa selección por torneo binario, cruce genético de un punto, mutación gaussiana estocástica y elitismo estricto para conservar la mejor solución en cada iteración.Registro de Historial: Almacena el mejor puntaje de aptitud por generación para permitir un análisis posterior de convergencia.Visualización 3D Interactiva (Plotly): Genera un globo terráqueo estilizado con tema espacial oscuro (fondo negro, letras blancas y cian de alta visibilidad) donde se aprecian con exactitud las regiones, satélites, zonas de exclusión y obstáculos.

# **Entregable 3. Pruebas y validación**

```python
# ==========================================
# ENTREGABLE 3: PRUEBAS Y VALIDACIÓN AUTOMATIZADA
# ==========================================

print("=== INICIANDO BATERÍA DE PRUEBAS ASTROSHIELD ===\n")

# --- Prueba 1: Escenario Inicial y Cobertura ---
print("[PRUEBA 1] Evaluando configuración inicial aleatoria...")
random_ind = optimizer.initialize_population()[0]
init_fit, init_regs, init_pen = evaluate_individual(random_ind)
print(f" -> Fitness inicial sin optimizar: {init_fit:.2f} (Penalización: {init_pen})")

# --- Prueba 2: Evolución de la Calidad ---
print("\n[PRUEBA 2] Analizando evolución a lo largo de las generaciones...")
print(f" -> Fitness inicial (Gen 0): {fitness_history[0]:.2f}")
print(f" -> Fitness intermedio (Gen 60): {fitness_history[60]:.2f}")
print(f" -> Fitness final (Gen 120): {fitness_history[-1]:.2f} (Mejora del {(fitness_history[-1]-fitness_history[0])/abs(fitness_history[0])*100:.1f}%)")

# --- Prueba 3 & 4: Zonas Restringidas y Posiciones Ocupadas ---
print("\n[PRUEBA 3 & 4] Verificando cumplimiento de restricciones espaciales...")
total_debris_violations = 0
total_blocked_violations = 0

for i, sat in enumerate(best_constellation):
    lat, lon = sat
    for deb in DEBRIS_ZONES:
        if angular_distance(lat, lon, deb["lat"], deb["lon"]) <= deb["radius_deg"]:
            total_debris_violations += 1
    for obj in BLOCKED_OBJECTS:
        if angular_distance(lat, lon, obj["lat"], obj["lon"]) <= obj["radius_deg"]:
            total_blocked_violations += 1

print(f" -> Violaciones en Zonas de Desechos: {total_debris_violations} (Esperado: 0)")
print(f" -> Violaciones en Objetos Ocupados (A/B): {total_blocked_violations} (Esperado: 0)")

# --- Prueba 5: Espacio Reducido / Condición Difícil ---
print("\n[PRUEBA 5 & 6] Simulando condición estricta / espacio reducido...")
# Simulamos expandiendo el radio de exclusión de los objetos al doble para medir resistencia
strict_violations = 0
for sat in best_constellation:
    for obj in BLOCKED_OBJECTS:
        if angular_distance(sat[0], sat[1], obj["lat"], obj["lon"]) <= (obj["radius_deg"] * 2):
            strict_violations += 1
print(f" -> Infracciones bajo radio de exclusión duplicado: {strict_violations}")

# --- Prueba 7: Variación de la Ejecución (Estabilidad) ---
print("\n[PRUEBA 7] Evaluando variabilidad entre múltiples ejecuciones...")
run_results = []
for run in range(5):
    opt_test = EvolutionaryConstellationOptimizer(pop_size=40, generations=50, mutation_rate=0.25)
    _, best_f, _ = opt_test.optimize()
    run_results.append(best_f)

print(f" -> Resultados de Fitness en 5 ejecuciones independientes: {[round(r, 2) for r in run_results]}")
print(f" -> Promedio: {np.mean(run_results):.2f} | Desviación Estándar: {np.std(run_results):.2f}")

# --- Prueba Adicional Propia: Sensibilidad al Radio de Observación ---
print("\n[PRUEBA ADICIONAL PROPIA] Sensibilidad al radio de observación (200 km vs 400 km):")
def test_observation_radius(radius_km):
    global OBSERVATION_RADIUS_KM
    old_r = OBSERVATION_RADIUS_KM
    OBSERVATION_RADIUS_KM = radius_km
    _, regs, _ = evaluate_individual(best_constellation)
    covered_count = sum(regs)
    OBSERVATION_RADIUS_KM = old_r
    return covered_count

cov_200 = test_observation_radius(200.0)
cov_400 = test_observation_radius(400.0)
print(f" -> Regiones cubiertas con 200 km de radio: {cov_200}/6")
print(f" -> Regiones cubiertas con 400 km de radio: {cov_400}/6")
print("\n=== PRUEBAS FINALIZADAS CON ÉXITO ===")
```

# **Entregable 5. Visualización de la constelación en 3D**

```python
import plotly.graph_objects as go
import numpy as np

def plot_interactive_earth(constellation):
    # Función auxiliar para convertir Lat/Lon a Cartesianas 3D
    def latlon_to_cartesian(lat, lon, radius=EARTH_RADIUS_KM):
        rad_lat = np.radians(lat)
        rad_lon = np.radians(lon)
        x = radius * np.cos(rad_lat) * np.cos(rad_lon)
        y = radius * np.cos(rad_lat) * np.sin(rad_lon)
        z = radius * np.sin(rad_lat)
        return x, y, z

    data = []

    # 1. Crear la esfera de la Tierra (con un tono azulado traslúcido para destacar en el espacio)
    u = np.linspace(0, 2 * np.pi, 50)
    v = np.linspace(0, np.pi, 50)
    xe = EARTH_RADIUS_KM * np.outer(np.cos(u), np.sin(v))
    ye = EARTH_RADIUS_KM * np.outer(np.sin(u), np.sin(v))
    ze = EARTH_RADIUS_KM * np.outer(np.ones(np.size(u)), np.cos(v))

    earth_surface = go.Surface(
        x=xe, y=ye, z=ze,
        colorscale=[[0, 'rgba(0, 119, 190, 0.25)'], [1, 'rgba(0, 119, 190, 0.25)']],
        showscale=False,
        name="Tierra"
    )
    data.append(earth_surface)

    # 2. Regiones de Interés (Texto blanco brillante)
    reg_x, reg_y, reg_z, reg_texts = [], [], [], []
    for reg in REGIONS:
        rx, ry, rz = latlon_to_cartesian(reg["lat"], reg["lon"])
        reg_x.append(rx); reg_y.append(ry); reg_z.append(rz)
        reg_texts.append(f"<b>{reg['name']}</b><br>Prioridad: {reg['priority']}")

    data.append(go.Scatter3d(
        x=reg_x, y=reg_y, z=reg_z,
        mode='markers+text',
        marker=dict(size=7, color='limegreen', symbol='circle'),
        text=[r['name'] for r in REGIONS],
        textposition="top center",
        textfont=dict(color='white', size=10),
        hovertext=reg_texts,
        hoverinfo="text",
        name="Regiones de Interés"
    ))

    # 3. Zonas de Desechos Espaciales
    deb_x, deb_y, deb_z, deb_texts = [], [], [], []
    for debris in DEBRIS_ZONES:
        dx, dy, dz = latlon_to_cartesian(debris["lat"], debris["lon"])
        deb_x.append(dx); deb_y.append(dy); deb_z.append(dz)
        deb_texts.append(f"<b>Desechos: {debris['name']}</b><br>Radio: {debris['radius_deg']}°")

    data.append(go.Scatter3d(
        x=deb_x, y=deb_y, z=deb_z,
        mode='markers+text',
        marker=dict(size=9, color='red', symbol='x'),
        text=[d['name'] for d in DEBRIS_ZONES],
        textposition="top center",
        textfont=dict(color='white', size=10),
        hovertext=deb_texts,
        hoverinfo="text",
        name="Zonas de Desechos"
    ))

    # 4. Objetos Ocupados
    obj_x, obj_y, obj_z, obj_texts = [], [], [], []
    for obj in BLOCKED_OBJECTS:
        ox, oy, oz = latlon_to_cartesian(obj["lat"], obj["lon"])
        obj_x.append(ox); obj_y.append(oy); obj_z.append(oz)
        obj_texts.append(f"<b>{obj['name']}</b><br>Radio de exclusión: {obj['radius_deg']}°")

    data.append(go.Scatter3d(
        x=obj_x, y=obj_y, z=obj_z,
        mode='markers+text',
        marker=dict(size=9, color='orange', symbol='diamond'),
        text=[o['name'] for o in BLOCKED_OBJECTS],
        textposition="top center",
        textfont=dict(color='white', size=10),
        hovertext=obj_texts,
        hoverinfo="text",
        name="Objetos Ocupados"
    ))

    # 5. Satélites de la Constelación (Órbita a +400 km)
    sat_x, sat_y, sat_z, sat_texts = [], [], [], []
    for i, sat in enumerate(constellation):
        sx, sy, sz = latlon_to_cartesian(sat[0], sat[1], radius=EARTH_RADIUS_KM + 400)
        sat_x.append(sx); sat_y.append(sy); sat_z.append(sz)
        sat_texts.append(f"<b>Satélite {i+1}</b><br>Lat: {sat[0]:.2f}°<br>Lon: {sat[1]:.2f}°")

    data.append(go.Scatter3d(
        x=sat_x, y=sat_y, z=sat_z,
        mode='markers+text',
        marker=dict(size=10, color='cyan', symbol='square'),
        text=[f"Sat-{i+1}" for i in range(len(constellation))],
        textposition="top center",
        textfont=dict(color='cyan', size=11, family='Arial Black'),
        hovertext=sat_texts,
        hoverinfo="text",
        name="Satélites (AstroShield)"
    ))

    # Configuración del diseño interactivo con fondo negro total y letras blancas
    layout = go.Layout(
        title=dict(text="<b>AstroShield — Globo Terráqueo 3D Interactivo</b>", font=dict(color='white', size=16)),
        paper_bgcolor='black',
        plot_bgcolor='black',
        scene=dict(
            xaxis=dict(title='X (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            yaxis=dict(title='Y (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            zaxis=dict(title='Z (km)', titlefont=dict(color='white'), tickfont=dict(color='white'), backgroundcolor="black", gridcolor="gray"),
            aspectmode='cube' # Mantiene la proporción esférica perfecta
        ),
        legend=dict(x=0, y=0.9, font=dict(color='white'), bgcolor='rgba(0,0,0,0.5)'),
        margin=dict(l=0, r=0, b=0, t=40)
    )

    fig = go.Figure(data=data, layout=layout)
    fig.show()

# Ejecutar la visualización interactiva
plot_interactive_earth(best_constellation)
```
